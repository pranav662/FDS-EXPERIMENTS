#**Experiment NO. 4 : Explore the dataset using descriptive statistics to gain insights into the data distribution and identify patterns.**

**Theory**

Descriptive statistics is the process of collecting, organizing, summarizing, and presenting data in a meaningful way. It helps us understand the main characteristics of a dataset without making predictions or generalizations beyond the available data.

When a dataset contains a large number of observations, it is difficult to understand the data by looking at every individual value. Descriptive statistics provides a compact summary that helps identify:

*   The central or typical value of the data
*   How widely the values are spread
*   Minimum and maximum values
*   Distribution of the data
*   Possible unusual or extreme values
*   Patterns and relationships within the dataset

**Software Requirements**

Python 3.x

Jupyter Notebook / Google Colab

Pandas

NumPy



```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

# Basic Information
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

# Descriptive Statistics
print("\nDescriptive Statistics:")
print(df.describe())

# Age Statistics
print("\nAge Statistics")

print("Mean:", df["Age"].mean())
print("Median:", df["Age"].median())
print("Mode:", df["Age"].mode()[0])
print("Minimum:", df["Age"].min())
print("Maximum:", df["Age"].max())
print("Range:", df["Age"].max() - df["Age"].min())
print("Variance:", df["Age"].var())
print("Standard Deviation:", df["Age"].std())

# Fare Statistics
print("\nFare Statistics")

print("Mean:", df["Fare"].mean())
print("Median:", df["Fare"].median())
print("Minimum:", df["Fare"].min())
print("Maximum:", df["Fare"].max())

# Categorical Analysis
print("\nGender Distribution:")
print(df["Sex"].value_counts())

print("\nPassenger Class Distribution:")
print(df["Pclass"].value_counts())

# Group Analysis
print("\nAverage Age by Passenger Class:")
print(df.groupby("Pclass")["Age"].mean())

print("\nAverage Fare by Passenger Class:")
print(df.groupby("Pclass")["Fare"].mean())

# Survival Rate
print("\nSurvival Rate by Passenger Class:")
print(df.groupby("Pclass")["Survived"].mean())

# Histogram
plt.figure(figsize=(8,5))

plt.hist(df["Age"].dropna(), bins=20)

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")

plt.show()

# Box Plot
plt.figure(figsize=(8,5))

plt.boxplot(df["Fare"].dropna())

plt.ylabel("Fare")
plt.title("Fare Distribution")

plt.show()
```

#**Experiment NO. 5 :  For given dataset, measure central tendency and variance. Data analysis in terms of Normal Distribution,Binomial Distribution.**



**Theory**

Statistical data analysis is the process of collecting,
organizing, summarizing, and interpreting data to understand its characteristics and patterns. In this experiment, a dataset containing student marks is analyzed using measures of central tendency, measures of dispersion, Normal Distribution, and Binomial Distribution.

The main purpose is to understand how the marks are distributed, how much they vary from the average, and how probability distributions can be used to analyze data.

**Software Requirements**
*   Python 3.x
*   Jupyter Notebook / Google Colab
*   Pandas
*   NumPy
*   Matplotlib
*   SciPy


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, binom

# Dataset
marks = [65, 70, 72, 68, 75, 80, 72, 85, 78, 70,
         82, 75, 68, 90, 72]

df = pd.DataFrame({"Marks": marks})

# Central Tendency
print("Mean =", df["Marks"].mean())
print("Median =", df["Marks"].median())
print("Mode =", df["Marks"].mode().values)

# Variance and Standard Deviation
print("Variance =", df["Marks"].var())
print("Standard Deviation =", df["Marks"].std())

# Normal Distribution
mean = df["Marks"].mean()
std = df["Marks"].std()

x = np.linspace(df["Marks"].min(), df["Marks"].max(), 100)
normal_curve = norm.pdf(x, mean, std)

plt.figure(figsize=(8,5))
plt.hist(df["Marks"], bins=6, density=True)
plt.plot(x, normal_curve)
plt.xlabel("Marks")
plt.ylabel("Density")
plt.title("Normal Distribution of Student Marks")
plt.show()

# Probability using Normal Distribution
prob_normal = 1 - norm.cdf(80, mean, std)
print("Probability of marks greater than 80 =", prob_normal)

# Binomial Distribution
n = 10
p = 0.7
x = 7

prob_binomial = binom.pmf(x, n, p)

print("Probability of exactly 7 successes =", prob_binomial)
```
